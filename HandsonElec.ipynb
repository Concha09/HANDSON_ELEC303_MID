```python
import matplotlib.pyplot as plt

crime['Date'] = pd.to_datetime(crime['Date'])

crime['Year'] = crime['Date'].dt.year
crime['Month'] = crime['Date'].dt.month

monthly_crimes = crime.groupby(['Year', 'Month']).size()

monthly_crimes = monthly_crimes.unstack(level=0)

monthly_crimes.plot(
    marker='o', 
    figsize=(10, 5),
    title='Monthly crime by Year (2001-2025)'
)
plt.xticks(range(1, 13), ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec'])
plt.legend(title='Yearly and Monthly Crimes Comparison', bbox_to_anchor = (1.05,1), loc='upper left')
plt.grid(alpha=0.3)
plt.show()
```

<h1>Insight 9: Across nearly all years,crime levels peak between May and July, Then gradually decline from September to December suggesting seasonality mid-year tends to be the most active crime period.</h1>
<h1>Insight 10: In year 2020 onward, lines seem closer together and slightly flatter, suggesting crime might be stabilizing or sometimes decreasing in recent years compared to earlier 2000s. </h1>
<h1>Insight 11: Although the general pattern repeats each year, the height of peaks differs. </h1>

```python
crime_counts = crime['Primary Type'].value_counts()

crimes_2001 = crime[crime['Year'] == 2001]

tc_2001 = crimes_2001['Primary Type'].value_counts()

plt.pie(tc_2001,
        labels = tc_2001.index,
        autopct = '%1.1f%%',
        textprops={'fontsize':6},
        # startangle = 90 
       )
plt.gcf().set_size_inches(10,7)
plt.title('Crimes in Chicago 2001')
plt.show()
```
